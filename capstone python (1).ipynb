```python
# Import the pandas library to work with data tables, and name it 'pd' for short
import pandas as pd

# 1. Create the Project Data Table using a dictionary structure
project_data = {
    'ID': ['A001', 'A002', 'A003', 'A004', 'A005', 'A002', 'A005', 'A003', 'A001', 'A003', 'A001', 'A004', 'A004', 'A005'],
    'Project': ['Project 1', 'Project 2', 'Project 3', 'Project 4', 'Project 5', 'Project 6', 'Project 7', 'Project 8', 'Project 9', 'Project 10', 'Project 11', 'Project 12', 'Project 13', 'Project 14'],
    'Cost': [1002000, 2000000, 4500000, 5500000, None, 6800000, 4000000, 3500000, None, 3000000, 2000000, 1000000, 3000000, 2000000],
    'Status': ['Finished', 'Ongoing', 'Finished', 'Ongoing', 'Finished', 'Failed', 'Finished', 'Failed', 'Ongoing', 'Finished', 'Failed', 'Ongoing', 'Finished', 'Finished']
}
# Convert the project dictionary into a structured Pandas DataFrame
df_project = pd.DataFrame(project_data)

# 2. Create the Employee Data Table using a dictionary structure
employee_data = {
    'ID': ['A001', 'A002', 'A003', 'A004', 'A005'],
    'Name': ['John Alter', 'Alice Luxumberg', 'Tom Sabestine', 'Nina Adgra', 'Amy Johny'],
    'Gender': ['M', 'F', 'M', 'F', 'F'],
    'City': ['Paris', 'London', 'Berlin', 'Newyork', 'Madrid'],
    'Age': [25, 27, 29, 31, 30]
}
# Convert the employee dictionary into a structured Pandas DataFrame
df_employee = pd.DataFrame(employee_data)

# 3. Create the Seniority Level Data Table using a dictionary structure
seniority_data = {
    'ID': ['A001', 'A002', 'A003', 'A004', 'A005'],
    'Designation Level': [2, 2, 3, 2, 3]
}
# Convert the seniority dictionary into a structured Pandas DataFrame
df_seniority = pd.DataFrame(seniority_data)

# Save the Project DataFrame to a CSV file without including the row index numbers
df_project.to_csv('project.csv', index=False)
# Save the Employee DataFrame to a CSV file without including the row index numbers
df_employee.to_csv('employee.csv', index=False)
# Save the Seniority DataFrame to a CSV file without including the row index numbers
df_seniority.to_csv('seniority.csv', index=False)

# Display the Project DataFrame to confirm everything looks exactly like the PDF
df_project
```

```python
# Import the numpy library to help identify missing data markers
import numpy as np

# --- TASK 2: FILL MISSING VALUES ---
# Load the project dataset we saved in Task 1 to make sure we use the saved file
df_project = pd.read_csv('project.csv')

# Calculate the mean (average) of the 'Cost' column, ignoring any empty/missing values
running_average = df_project['Cost'].mean()

# Use a FOR loop to iterate through every row index from 0 to the length of the dataframe
for index in range(len(df_project)):
    # Check IF the cost value at the current row index is missing (null/NaN)
    if pd.isna(df_project.loc[index, 'Cost']):
        # If it is missing, replace it with the calculated running average cost
        df_project.loc[index, 'Cost'] = running_average

# Save the updated project dataframe back to a CSV file to preserve our changes
df_project.to_csv('project.csv', index=False)


# --- TASK 3: SPLIT EMPLOYEE NAMES ---
# Load the employee dataset we saved in Task 1 to ensure we work on the latest file
df_employee = pd.read_csv('employee.csv')

# Split the 'Name' column by the space character to extract the first name part
df_employee['First Name'] = df_employee['Name'].str.split(' ').str[0]

# Split the 'Name' column by the space character to extract the last name part
df_employee['Last Name'] = df_employee['Name'].str.split(' ').str[1]

# Drop the original 'Name' column since we now have it separated into two columns
df_employee = df_employee.drop(columns=['Name'])

# Save the updated employee dataframe back to its CSV file to save the split names
df_employee.to_csv('employee.csv', index=False)

# Display the final updated employee dataframe to verify the new columns look correct
df_employee
```

```python
# --- TASK 4: JOIN ALL THREE DATAFRAMES INTO A SINGLE DATAFRAME NAMED "Final" ---

# Step 1: Read the latest versions of all three saved CSV files
df_project = pd.read_csv('project.csv')
df_employee = pd.read_csv('employee.csv')
df_seniority = pd.read_csv('seniority.csv')  # Note: ensure this matches the filename you used in Task 1

# Step 2: Merge the project dataframe with the employee dataframe on the 'ID' column
df_intermediate = pd.merge(df_project, df_employee, on='ID')

# Step 3: Merge the resulting dataframe with the seniority dataframe on the 'ID' column to create 'Final'
Final = pd.merge(df_intermediate, df_seniority, on='ID')

# Step 4: Save the combined 'Final' dataframe into a new CSV file as required by the setup
Final.to_csv('final.csv', index=False)

# Step 5: Display the Final dataframe to inspect the merged columns
Final
```

```python
# 1. Initialize the Bonus column with 0 for all rows
Final['Bonus'] = 0.0

# 2. Calculate the 5% bonus only for rows where the Status is 'Finished'
Final.loc[Final['Status'] == 'Finished', 'Bonus'] = Final['Cost'] * 0.05

# 3. View the updated DataFrame to verify the new column
Final
```

```python
# Part 1: Subtract 1 from Designation Level for projects that 'Failed'
# Note: In your table, the status is capitalized as 'Failed'
Final.loc[Final['Status'] == 'Failed', 'Designation Level'] -= 1

# Part 2: Keep only rows where Designation Level is 4 or below (deleting anything above 4)
Final = Final[Final['Designation Level'] <= 4]

# View the updated DataFrame
Final
```

```python
# 1. Add "Mr. " to the First Name if Gender is 'M'
Final.loc[Final['Gender'] == 'M', 'First Name'] = 'Mr. ' + Final['First Name']

# 2. Add "Mrs. " to the First Name if Gender is 'F'
Final.loc[Final['Gender'] == 'F', 'First Name'] = 'Mrs. ' + Final['First Name']

# 3. Drop the Gender column
Final = Final.drop(columns=['Gender'])

# View the updated DataFrame
Final
```

```python
# Apply a condition across rows: if Age is greater than 29, add 1 to Designation Level, otherwise keep it the same
Final['Designation Level'] = Final.apply(lambda row: row['Designation Level'] + 1 if row['Age'] > 29 else row['Designation Level'], axis=1)

# Display the updated Final dataframe to verify the promotions
Final
```

```python
# Group by ID and First Name, then calculate the sum of the Cost column for each unique employee
TotalProjCost = Final.groupby(['ID', 'First Name'])['Cost'].sum().reset_index()

# Rename the aggregated column from 'Cost' to 'Total cost' to match the task requirements
TotalProjCost = TotalProjCost.rename(columns={'Cost': 'Total cost'})

# Display the newly created TotalProjCost dataframe
TotalProjCost
```

```python
# Filter the dataframe for rows where the 'City' column contains the letter 'o' (case-insensitive)
employees_with_o = Final[Final['City'].str.contains('o', case=False)]

# Display the filtered employees as a clean table
employees_with_o
```
